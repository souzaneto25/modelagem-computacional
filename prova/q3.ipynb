```python
import sympy as sp

def lagrange(x_nos, y_nos, x):
    n = len(x_nos)
    p = 0.0
    for i in range(n):
        L = 1.0
        for j in range(n):
            if j != i:
                L *= (x - x_nos[j]) / (x_nos[i] - x_nos[j])
        p += y_nos[i] * L
    return p

x = sp.Symbol('x')
x_nos = [0.5, 1.5, 3.5, 6.0]
y_nos = [0.2, 0.4, 0.6, 0.8]

n = len(x_nos)
P = 0
for i in range(n):
    L = 1
    for j in range(n):
        if j != i:
            L *= (x - x_nos[j]) / (x_nos[i] - x_nos[j])
    P += y_nos[i] * L

P_expandido = sp.expand(P)
print("Polinômio interpolador de Lagrange (grau 3):")
sp.pprint(P_expandido)
          
I_estimado = lagrange(x_nos, y_nos, 0.5)
print("=== Estimativa da corrente ===")
print(f"I(0.5 V) estimado = {I_estimado:.4f} mA")

I_real = 2.6  # mA
erro_absoluto = abs(I_estimado - I_real)
print(f"\n=== Erro absoluto ===")
print(f"I real      = {I_real} mA")
print(f"I estimado  = {I_estimado:.4f} mA")
print(f"Erro        = {erro_absoluto:.4f} mA")
```
