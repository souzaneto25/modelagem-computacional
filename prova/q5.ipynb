```python
import numpy as np

def runge_kutta_4th_order(f, x0, y0, h, x_final):

    x_values = [x0]
    y_values = [y0]

    x = x0
    y = y0

    while x < x_final:
        k1 = f(x, y)
        k2 = f(x + h/2, y + h/2 * k1)
        k3 = f(x + h/2, y + h/2 * k2)
        k4 = f(x + h, y + h * k3)

        y = y + (h/6) * (k1 + 2*k2 + 2*k3 + k4)
        x = x + h

        x_values.append(x)
        y_values.append(y)

    return np.array(x_values), np.array(y_values)
```

```python
def f(x, u):
    return np.array([u[1], -4*u[1] - 5*u[0] + 10*np.cos(2*x)])

C1 = -2/13
C2 = -36/13
def y_exato(x):
    return (np.exp(-2*x) * (C1*np.cos(x) + C2*np.sin(x))
            + 2/13*np.cos(2*x) + 16/13*np.sin(2*x))

xs, us = runge_kutta_4th_order(f, 0, np.array([0.0, 0.0]), 0.1, 4.0)
y_num = us[:, 0]
y_ex  = y_exato(xs)
erros = np.abs(y_num - y_ex)

print('tababela de valores')
print(f'{"x":<8} {"y RK4":<14} {"y exato":<14} {"erro abs"}')
for i in range(len(xs)):
    print(f'{xs[i]:<8.2f} {y_num[i]:<14.6f} {y_ex[i]:<14.6f} {erros[i]:.3e}')

print(f'\nErro máximo: {erros.max():.4e}')

#Obter x(1.5) e x'(1.5)
x_target = 1.5
index = np.where(np.isclose(xs, x_target))[0][0]
x_1_5 = xs[index]
y_1_5 = y_num[index]
y_prime_ord_1_5 = us[index, 1]
print(f'\nValor de x(1.5): {y_1_5:.6f}')
print(f'Valor de x\'(1.5): {y_prime_ord_1_5:.6f}')
```
